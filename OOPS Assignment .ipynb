7- What is polymorphism in OOP ?

- Polymorphism is one of the core concepts of Object-Oriented Programming (OOP).
The word polymorphism means “many forms.”

    In OOP, it allows the same method or operation to behave differently depending on the object that is calling it.

    Simple Definition

    Polymorphism allows different classes to define the same method name but implement it differently.

    Why Polymorphism Is Useful

    Increases flexibility
    
    Allows code to work with different objects in a uniform way
    
    Supports method overriding and method overloading (though Python mainly supports overriding)

    Types of Polymorphism

    Compile-Time Polymorphism (Method Overloading)

    Same method name with different number of parameters

    Not naturally supported in Python, but can be mimicked

    Run-Time Polymorphism (Method Overriding)

    Child class provides its own implementation of a method from the parent class

    Common in Python

    Example of Polymorphism (Method Overriding)

    class Animal:

    def speak(self):
        return "Animal makes a sound"

    class Dog(Animal):
    
    def speak(self):
        return "Dog barks"

    class Cat(Animal):
    
    def speak(self):
        return "Cat meows"

    # Polymorphism in action
    
    animals = [Dog(), Cat(), Animal()]

    for a in animals:
    
    print(a.speak())

    Dog barks
    
    Cat meows

    Animal makes a sound

    Even though the method name is the same (speak()), the behavior changes depending on the object → this is polymorphism.

    Another Example: Same Function, Different Types

    print(len("Hello"))   # 5
    
    print(len([1, 2, 3])) # 3
    
    print(len({"a": 1}))  # 1

    The function len() behaves differently based on the object — this is also polymorphism.

8- How is encapsulation achieved in Python ?

- Encapsulation in Python is the OOP concept of bundling data (variables) and methods (functions) inside a class and restricting direct access to some of the class internal components.
This helps protect data from accidental modification and keeps the code organized and secure.

    Python achieves encapsulation using three types of access modifiers:

    I- Public Members

    Can be accessed from anywhere.

    Defined normally.
    
    class MyClass:
   
    def __init__(self):
        self.name = "Sunny"   # public

    II- Protected Members (_single_underscore)

    Meant for internal use within the class and subclasses.

    Not strictly enforced by Python but treated as protected by convention.

    class MyClass:
    
    def __init__(self):
        self._balance = 5000  # protected (convention)

    III- Private Members (__double_underscore)

    Name-mangled by Python to prevent direct access from outside.

    Best used for sensitive data.

    class MyClass:
    
    def __init__(self):
        self.__password = "12345"  # private

    Trying to access it directly causes an error:

    obj = MyClass()
    
    print(obj.__password)  # ❌ AttributeError

    Access is still possible internally using getters/setters.

    Encapsulation Example (Using Getter and Setter)

    class BankAccount:
    
    def __init__(self, owner, balance):
        self.owner = owner
        self.__balance = balance   # private variable

    # getter method
    
    def get_balance(self):
        return self.__balance

    # setter method
    
    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount

    account = BankAccount("Sunny", 1000)

    print(account.get_balance())   # ✔ 1000

    account.deposit(500)

    print(account.get_balance())   # ✔ 1500

10- What are class and static methods in Python ?

- In Python, class methods and static methods are special types of methods inside a class that behave differently from instance methods.

    Class Method

    A class method works with the class itself, not with individual objects.

    Key Points

    . instance (self)

    - Key Points

    Defined using @staticmethod decorator.

    Does NOT take self or cls.

    Does not access instance or class data.

    Used for utility/helper functions.

    Example of a Static Method:

    class MathOps:
    
    @staticmethod
    
    def add(a, b):

11- What is method overloading in Python ?

- Method overloading in Python refers to defining multiple methods with the       same name but different parameters—a feature that exists in languages like Java or C++.

    However, Python does NOT support method overloading in the traditional sense.

    If you define multiple methods with the same name in a class, the latest definition overwrites the previous ones.

    - Using Default Parameters (Most Common)

    class Example:
    def greet(self, name=None):
        if name:
            return f"Hello, {name}!"
        else:
            return "Hello!"
    
    - Using *args and **kwargs

    class Calculator:

    def add(self, *numbers):

        return sum(numbers)

    - Using functools.singledispatch (Type-based Overloading)

    Python allows function overloading based on argument type.
    
    from functools import singledispatch

    @singledispatch

    def show(value):

    print("Default:", value)

    @show.register

    def _(value: int):

    print("Integer:", value)

    @show.register

    def _(value: list):

    print("List:", value)

    show(10)      # Integer

    show([1, 2])  # List

13- What is a property decorator in Python ?

- A property decorator (@property) in Python is a built-in decorator used to convert a class method into an attribute-like access, allowing you to use methods as if they were variables.

    It helps implement getter, setter, and deleter functionality in a clean, Pythonic way.

    class Student:

    def __init__(self, marks):

        self._marks = marks

    @property

    def marks(self):

        return self._marks

        s = Student(90)

    print(s.marks)  # accessed like an attribute, but it's a method

    Adding a Setter (to modify value safely)

    class Student:

    def __init__(self, marks):

        self._marks = marks

    @property

    def marks(self):

        return self._marks

    @marks.setter

    def marks(self, value):

        if value < 0:
            raise ValueError("Marks cannot be negative!")
        self._marks = value

    s = Student(80)

    s.marks = 95  # setter runs automatically

23- How does polymorphism work in Python with inheritance ?

When you call that method using a parent-class reference, Python automatically decides at runtime which version of the method to execute.

    This is known as runtime polymorphism or method overriding.

    How It Works (Step-by-step)

    I-Parent class defines a method.

    II-Child class overrides (redefines) the same method.

    III-Python determines at runtime which method to call based on the object.

    IV-Same interface → different behavior.

    Example: Polymorphism via Method Overriding -

    class Animal:

    def sound(self):

        return "Some sound"

    class Dog(Animal):
    
    def sound(self):

        return "Bark"

    class Cat(Animal):

    def sound(self):

        return "Meow"

    pets = [Dog(), Cat(), Animal()]

    for pet in pets:

    print(pet.sound())  

    Bark

    Meow
    
    Some sound

    Even though each object uses the same .sound() method name,
    
    the behavior changes depending on the object type — that is polymorphism.

25- What is the purpose of the __call__ method in Python ?

- The __call__ method in Python is a special (dunder) method that allows an object of a class to be called like a function.

    Simple Definition (Interview-ready)

    The __call__ method makes an instance callable, meaning you can use parentheses () on the object to execute custom behavior defined in the method.

    Purpose of __call__

    -Treat objects like functions for flexibility.

    -Useful for wrappers, decorators, or callable classes.

    -Enables stateful functions: objects that hold data and behave like functions.

    Example of __call__ -

    class Greeting:

    def __init__(self, name):

        self.name = name

    def __call__(self, message):
    
        print(f"{message}, {self.name}!")

    greet = Greeting("Sunny")

    greet("Hello")  # Using the object like a function

    Hello, Sunny!

    greet is an object

    greet("Hello") internally calls greet.__call__("Hello")

        if amount <= self.__balance:

        return self.__balance   # only accessible through this method

    print("Current Balance:", account.check_balance())
